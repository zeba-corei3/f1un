```python
#number of drivers
D = 3

#driver name - encoding
driver_id = {0: 'Lando Norris', 1: 'Max Verstappen', 2: 'Oscar Piastri', '-':'-'}

#Points available (0 depicts out of points or DNF or DSQ)
points = np.array([25, 18, 15, 12, 10, 8, 6, 4, 2, 1, 0])

#Ranks they can finish at
ranks = np.array([i for i in range(0, 12)])
```

```python
import itertools

def seat_arrangements(n, r):
    people = list(range(0, r))
    seats = list(range(n))

    arrangements = []

    # Step 1: choose r seat positions out of n
    for chosen_seats in itertools.combinations(seats, r):
        
        # Step 2: permute the r people in these seats
        for perm in itertools.permutations(people, r):
            
            # Build matrix row (length = n)
            row = ['-' for _ in range(n)]   # '-' represents empty seat
            for seat_index, person in zip(chosen_seats, perm):
                row[seat_index] = person

            arrangements.append(row)

    return arrangements


# Example
n = 11
r = 3
result = seat_arrangements(n, r)

for i in range(10, 20):
    print(result[i], i)
print("Total arrangements:", len(result))
```

```python
def positions(poss):
    top_three_positions = np.empty((len(poss), 3), dtype=int)
    for i in range(len(poss)):
        top_three_positions[i][0] = poss[i].index(0)
        top_three_positions[i][1] = poss[i].index(1)
        top_three_positions[i][2] = poss[i].index(2)

    return top_three_positions

def allot_points(driver, standings, rank):
    if driver=='-':
        pass
    else:
        # print(f'Driver: {driver} receives {points[rank]}')
        standings[driver] += points[rank]
        # print(f'Points Now = {standings[driver]}')
        return standings

def print_results(pos_i, standings, top_three_p):
    """
       for every list of possible finishes,
       print points table for the top 3
       get the maximum no of points from standings list 
       print the name of the driver who scored the most.
    """
    # for i in range(3):
    print("\t\t\tPermuutation", i)
    print("Position | Driver | Points |")
    print("--------------------------")
    
    for p in standings:
            print(f'{top_three_p[pos_i][standings.index(p)]}  |  {driver_id[standings.index(p)]}  |  {p}')
        
    print("--------------------------")

    winner = standings.index(max(standings))
    print(f'\t\t{driver_id[winner]} WINS THE 2025 FORMULA 1 CHAMPIONSHIP')
        
    print("--------------------------")
```
