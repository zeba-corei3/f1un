```python
event = fastf1.get_session(2024, 8, 'Q')
event.load()
```

```python
event.results.columns
```

```python
event.results.loc[:, ['Abbreviation', 'Q1', 'Q2', 'Q3']]
```

```python
event.load(telemetry = True)
```

```python
import matplotlib.pyplot as plt
import fastf1.plotting as fp
```

```python
fp.setup_mpl(color_scheme='fastf1')

lec_fast = event.laps.pick_drivers('LEC')
```

```python
fastestlap = lec_fast.pick_fastest()
lec_car = lec_fast.get_car_data()

t = lec_car['Time']
speed = lec_car['Speed']
```

```python
lec_fast
```

```python
plt.plot(t, speed, label='fast', linewidth=0.5)
# plt.legend()
plt.show()
```

```python
monza = fastf1.get_session(2019, 'Monza', 'Q')
monza.load()
```
